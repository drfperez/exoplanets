```python
!wget "https://zenodo.org/records/10161743/files/ERS_DataSynthesis_Zenodo.zip?download=1" -O data.zip
!unzip -q data.zip -d wasp39b_data

import os
import pandas as pd
import matplotlib.pyplot as plt

# Buscar el fitxer de l'espectre de transmissió combinat
spectrum_file = None
for root, dirs, files in os.walk('wasp39b_data'):
    for file in files:
        if 'transmission' in file.lower() and ('combined' in file.lower() or 'benchmark' in file.lower()) and (file.endswith('.csv') or file.endswith('.txt') or file.endswith('.dat')):
            spectrum_file = os.path.join(root, file)
            break
    if spectrum_file:
        break

if spectrum_file:
    # Carregar les dades (assumint columnes: wavelength, depth o (Rp/Rs)^2, error; ajusta si cal)
    try:
        df = pd.read_csv(spectrum_file, comment='#', delim_whitespace=True, names=['wavelength', 'depth', 'error'])
    except:
        df = pd.read_csv(spectrum_file, comment='#', delim_whitespace=True)

    # Analitzar: traçar l'espectre
    plt.figure(figsize=(10, 6))
    plt.errorbar(df['wavelength'], df['depth'], yerr=df.get('error', None), fmt='o', label='Espectre de transmissió')
    plt.xlabel('Longitud d\'ona (μm)')
    plt.ylabel('(Rp/Rs)^2 o profunditat de trànsit')
    plt.title('Espectre de transmissió de WASP-39b (JWST)')
    plt.legend()
    plt.grid(True)
    plt.show()
else:
    print("No s'ha trobat un fitxer d'espectre de transmissió combinat. Comprova el directori 'wasp39b_data' per als fitxers extrets.")
```
